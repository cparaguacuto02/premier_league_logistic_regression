```python
#Numeric Features
df.describe()

#encode result in order to run correlation
le_temp = LabelEncoder()
df['result_encoded'] = le_temp.fit_transform(df['result'])

#feature and target correlation check
df.corr(numeric_only=True)['result_encoded'].sort_values()
```
